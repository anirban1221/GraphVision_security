```python
import pandas as pd
import numpy as np

# Columns we initially care about
transaction_cols = [
    "TransactionID",
    "isFraud",
    "TransactionDT",
    "TransactionAmt",
    "ProductCD",

    # Payment/card information
    "card1", "card2", "card3", "card4", "card5", "card6",

    # Address information
    "addr1", "addr2",

    # Email information
    "P_emaildomain",
    "R_emaildomain",
]

identity_cols = [
    "TransactionID",
    "DeviceType",
    "DeviceInfo",
] + [f"id_{i:02d}" for i in range(1, 39)]

tx = train_transaction[transaction_cols].copy()

identity = train_identity[identity_cols].copy()

df = tx.merge(
    identity,
    on="TransactionID",
    how="left"
)

print(df.shape)
```

```python
id_cols = [f"id_{i:02d}" for i in range(1, 39)]

report = []

for col in id_cols:
    s = df[col]

    vc = s.dropna().value_counts()

    report.append({
        "column": col,
        "non_null": s.notna().sum(),
        "missing_pct": round(s.isna().mean() * 100, 2),
        "unique": s.nunique(dropna=True),
        "values_repeated": (vc > 1).sum(),
        "max_frequency": vc.max() if len(vc) else 0
    })

identity_report = pd.DataFrame(report)

print(identity_report.to_string(index=False))
```

```python
graph_candidates = [
    "card1", "card2", "card3", "card4", "card5", "card6",
    "addr1", "addr2",
    "P_emaildomain", "R_emaildomain",
    "DeviceType", "DeviceInfo"
]

results = []

overall_fraud_rate = df["isFraud"].mean()

for col in graph_candidates:

    temp = df[[col, "isFraud"]].dropna(subset=[col])

    stats = (
        temp.groupby(col)["isFraud"]
        .agg(
            transactions="count",
            frauds="sum",
            fraud_rate="mean"
        )
    )

    repeated = stats[stats["transactions"] >= 2]
    heavily_shared = stats[stats["transactions"] >= 10]

    results.append({
        "column": col,
        "non_null": len(temp),
        "missing_pct": round(df[col].isna().mean() * 100, 2),
        "unique": temp[col].nunique(),

        # Can this attribute connect multiple transactions?
        "repeated_values": len(repeated),
        "values_10plus_tx": len(heavily_shared),
        "max_transactions": int(stats["transactions"].max()),

        # Fraud relationship
        "overall_fraud_rate": round(overall_fraud_rate, 4),
        "max_fraud_rate": round(stats["fraud_rate"].max(), 4),
        "median_fraud_rate": round(stats["fraud_rate"].median(), 4)
    })

graph_report = pd.DataFrame(results)

print(graph_report.to_string(index=False))
```

```python
G = nx.Graph()

for _, row in edge_stats.iterrows():

    card_node = f"card_{int(row['card1'])}"
    device_node = f"device_{row['DeviceInfo']}"

    G.add_node(
        card_node,
        node_type="card"
    )

    G.add_node(
        device_node,
        node_type="device"
    )

    G.add_edge(
        card_node,
        device_node,
        transactions=int(row["transactions"]),
        frauds=int(row["frauds"]),
        fraud_rate=float(row["fraud_rate"])
    )

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
```

```python
import pandas as pd
import numpy as np

# --------------------------------------------------
# Candidate entity features
# --------------------------------------------------

candidate_features = [
    "card2", "card3", "card4", "card5", "card6",
    "DeviceInfo", "DeviceType",
    "addr1", "addr2",
    "P_emaildomain", "R_emaildomain"
]

candidate_features += [
    f"id_{i:02d}" for i in range(1, 39)
]

# Keep only columns that actually exist
candidate_features = [
    c for c in candidate_features if c in df.columns
]


# --------------------------------------------------
# Measure how much each feature connects cards
# --------------------------------------------------

results = []

for feature in candidate_features:

    temp = df[["card1", feature]].dropna()

    # Number of unique cards connected to each value
    value_card_counts = (
        temp.groupby(feature)["card1"]
        .nunique()
    )

    results.append({
        "feature": feature,
        "unique_values": temp[feature].nunique(),

        # How many values connect >= 2 cards?
        "multi_card_values":
            (value_card_counts >= 2).sum(),

        # How many values connect >= 5 cards?
        "5plus_card_values":
            (value_card_counts >= 5).sum(),

        # Typical number of cards connected by a value
        "median_cards":
            value_card_counts.median(),

        # 90th percentile
        "p90_cards":
            value_card_counts.quantile(0.90),

        # Largest connected group
        "max_cards":
            value_card_counts.max()
    })

entity_connectivity = (
    pd.DataFrame(results)
    .sort_values(
        "5plus_card_values",
        ascending=False
    )
)

print(entity_connectivity.to_string(index=False))
```

```python
specificity_results = []

for feature in candidate_features:

    temp = df[["card1", feature]].dropna()

    counts = (
        temp.groupby(feature)["card1"]
        .nunique()
    )

    specificity_results.append({
        "feature": feature,

        "unique_values": len(counts),

        # Fraction of values shared by multiple cards
        "multi_card_fraction":
            (counts >= 2).mean(),

        # Fraction of values shared by >= 5 cards
        "5plus_fraction":
            (counts >= 5).mean(),

        # Fraction of values that are extremely common
        "100plus_fraction":
            (counts >= 100).mean(),

        "max_cards": counts.max(),

        # Useful summary of typical sharing
        "median_cards": counts.median()
    })

specificity = (
    pd.DataFrame(specificity_results)
    .sort_values(
        "100plus_fraction",
        ascending=True
    )
)

print(specificity.to_string(index=False))
```

```python
graph_candidates = [
    "DeviceInfo",
    "id_19",
    "id_20",
    "id_31",
    "id_06",
    "id_30",
    "id_33",
    "id_05",
    "id_08",
    "id_17",
    "id_21",
    "addr1",
    "addr2",
    "card2",
    "card3",
    "card5"
]

graph_candidates = [
    c for c in graph_candidates
    if c in df.columns
]

print(graph_candidates)
```

```python
from itertools import combinations
from collections import defaultdict
import pandas as pd
import numpy as np

MAX_CARDS_PER_VALUE = 50

feature_edges = {}

for feature in graph_candidates:

    temp = (
        df[["card1", feature]]
        .dropna()
        .drop_duplicates()
    )

    # Group cards belonging to each entity value
    groups = temp.groupby(feature)["card1"]

    edges = set()

    for value, cards in groups:

        cards = list(cards)

        # Avoid huge generic hubs
        if len(cards) > MAX_CARDS_PER_VALUE:
            cards = cards[:MAX_CARDS_PER_VALUE]

        if len(cards) < 2:
            continue

        # Every pair of cards sharing this entity
        for a, b in combinations(sorted(cards), 2):
            edges.add((a, b))

    feature_edges[feature] = edges

    print(
        f"{feature:12s} -> "
        f"{len(edges):,} sampled card-pair edges"
    )
```

```python
features = list(feature_edges.keys())

redundancy_rows = []

for i in range(len(features)):

    for j in range(i + 1, len(features)):

        f1 = features[i]
        f2 = features[j]

        e1 = feature_edges[f1]
        e2 = feature_edges[f2]

        if not e1 or not e2:
            continue

        intersection = len(e1 & e2)
        union = len(e1 | e2)

        jaccard = intersection / union if union else 0

        redundancy_rows.append({
            "feature_1": f1,
            "feature_2": f2,
            "shared_edges": intersection,
            "jaccard": jaccard
        })

redundancy_df = (
    pd.DataFrame(redundancy_rows)
    .sort_values("jaccard", ascending=False)
)

print(
    redundancy_df
    .head(30)
    .to_string(index=False)
)
```

```python
card_stats = (
    train_df
    .groupby("card1")["isFraud"]
    .agg(
        card_transactions="size",
        card_frauds="sum"
    )
)

card_stats["card_fraud_rate"] = (
    card_stats["card_frauds"] /
    card_stats["card_transactions"]
)

print(card_stats.head())
```

```python
edge_quality_rows = []

BASELINE = train_df["isFraud"].mean()

for feature in graph_candidates:

    temp = train_df[
        ["card1", feature]
    ].dropna()

    # Entity-level transaction count
    entity_tx = (
        temp.groupby(feature)
        .size()
        .rename("entity_transactions")
    )

    # Keep reasonably observed entities
    mature_entities = entity_tx[
        entity_tx >= 5
    ].index

    temp = temp[
        temp[feature].isin(mature_entities)
    ]

    # Attach card fraud statistics
    temp = temp.join(
        card_stats["card_fraud_rate"],
        on="card1"
    )

    # Entity -> unique cards
    entity_card_stats = (
        temp
        .drop_duplicates([feature, "card1"])
        .groupby(feature)
        .agg(
            unique_cards=("card1", "nunique"),
            mean_card_fraud_rate=(
                "card_fraud_rate", "mean"
            ),
            median_card_fraud_rate=(
                "card_fraud_rate", "median"
            )
        )
    )

    if len(entity_card_stats) == 0:
        continue

    # How many entities have unusually
    # high fraud-connected cards?
    high_risk = (
        entity_card_stats[
            entity_card_stats["mean_card_fraud_rate"]
            >= BASELINE * 3
        ]
    )

    edge_quality_rows.append({
        "feature": feature,

        "mature_entities":
            len(entity_card_stats),

        "median_cards_per_entity":
            entity_card_stats["unique_cards"].median(),

        "median_card_fraud_rate":
            entity_card_stats[
                "mean_card_fraud_rate"
            ].median(),

        "p90_card_fraud_rate":
            entity_card_stats[
                "mean_card_fraud_rate"
            ].quantile(.90),

        "high_risk_entity_pct":
            len(high_risk) /
            len(entity_card_stats),

        "max_card_fraud_rate":
            entity_card_stats[
                "mean_card_fraud_rate"
            ].max()
    })

edge_quality = (
    pd.DataFrame(edge_quality_rows)
    .sort_values(
        "high_risk_entity_pct",
        ascending=False
    )
)

print(
    edge_quality.to_string(index=False)
)
```
